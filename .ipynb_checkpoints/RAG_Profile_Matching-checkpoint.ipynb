This notebook demonstrates:

3. Performance metrics: retrieval accuracy & latency
4. Qualitative analysis of top matches

## 1. Setup & Imports

```python
import sys, json, time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
from pathlib import Path

# Add project root to path
sys.path.insert(0, str(Path('').resolve()))

# Set plot style
plt.style.use('dark_background')
sns.set_palette('husl')
COLORS = ['#818cf8', '#c084fc', '#38bdf8', '#4ade80', '#fb923c', '#f87171']

print('All imports OK')
```

## 2. Initialize RAG System

```python
from resume_rag import ResumeRAG, load_all_resumes, extract_metadata

# Initialize the RAG pipeline
rag = ResumeRAG()
print('RAG system initialized')
print('Vector store chunks:', rag.store.count())
```

### 2.1 Index Resumes (if not already indexed)

```python
# Index resumes (skip if already indexed)
if rag.store.count() == 0:
    print('Indexing resumes...')
    rag.index_resumes()
else:
    print(f'Already indexed: {rag.store.count()} chunks in vector store')

stats = rag.get_store_stats()
for k, v in stats.items():
    print(f'  {k}: {v}')
```

## 3. Resume Dataset Analysis

```python
# Load all resumes and extract metadata
resumes = load_all_resumes()
print(f'Total resumes loaded: {len(resumes)}')

metadata_list = [extract_metadata(r) for r in resumes]
df = pd.DataFrame(metadata_list)
df.head()
```

### 3.1 Experience Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.patch.set_facecolor('#0d1117')

# Experience histogram
ax1 = axes[0]
ax1.set_facecolor('#141929')
ax1.hist(df['years_experience'], bins=10, color='#818cf8', edgecolor='#6366f1', alpha=0.85)
ax1.set_title('Years of Experience Distribution', color='white', fontsize=13, fontweight='bold')
ax1.set_xlabel('Years of Experience', color='#94a3b8')
ax1.set_ylabel('Count', color='#94a3b8')
ax1.tick_params(colors='#94a3b8')
for spine in ax1.spines.values(): spine.set_edgecolor('#374151')

# Top skills bar chart
ax2 = axes[1]
ax2.set_facecolor('#141929')
all_skills = [s for row in df['skills'] for s in (row if isinstance(row, list) else [])]
from collections import Counter
top_skills = Counter(all_skills).most_common(15)
skills_names = [s[0] for s in top_skills]
skills_counts = [s[1] for s in top_skills]
bars = ax2.barh(skills_names[::-1], skills_counts[::-1], color=COLORS[0], alpha=0.85)
ax2.set_title('Most Common Skills Across Resumes', color='white', fontsize=13, fontweight='bold')
ax2.set_xlabel('Frequency', color='#94a3b8')
ax2.tick_params(colors='#94a3b8')
for spine in ax2.spines.values(): spine.set_edgecolor('#374151')

plt.tight_layout()
plt.savefig('dataset_analysis.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()
print('Chart saved: dataset_analysis.png')
```

## 4. Job Matching Engine

```python
from job_matcher import JobMatcher, extract_skills_from_jd, extract_min_experience

matcher = JobMatcher()
print('JobMatcher initialized')
```

### 4.1 Test Query: Senior ML Engineer

```python
ML_JD = """
Senior Machine Learning Engineer - 5+ years experience
Required: Python, PyTorch, TensorFlow, MLflow, Kubeflow
LLMs, RAG, HuggingFace Transformers, NLP, BERT
AWS or GCP, Docker, Kubernetes
Vector databases: ChromaDB, Pinecone, Weaviate
"""

t0 = time.time()
result_ml = matcher.match(ML_JD, top_k=10)
latency_ml = time.time() - t0

print(f'Query latency: {latency_ml*1000:.0f}ms')
print(f'Detected skills: {result_ml["jd_skills"]}')
print(f'Min experience detected: {result_ml["min_experience_required"]} yrs')
print(f'Candidates evaluated: {result_ml["total_candidates_evaluated"]}')
print('\nTop 5 matches:')
for i, m in enumerate(result_ml['top_matches'][:5], 1):
    print(f'  {i}. {m["candidate_name"]:25s} Score={m["match_score"]:3d}/100  Exp={m["years_experience"]}yrs  Skills={m["matched_skills"][:3]}')
```

### 4.2 Run All Job Descriptions

```python
jd_dir = Path('./job_descriptions')
jd_files = list(jd_dir.glob('*.txt'))
print(f'Found {len(jd_files)} job descriptions')

all_results = {}
latencies = []

for jd_file in jd_files:
    jd_text = jd_file.read_text(encoding='utf-8')
    t0 = time.time()
    result = matcher.match(jd_text, top_k=10)
    elapsed_ms = (time.time() - t0) * 1000
    latencies.append(elapsed_ms)
    all_results[jd_file.stem] = result
    top = result['top_matches'][0] if result['top_matches'] else {}
    print(f'{jd_file.stem:30s} -> Top: {top.get("candidate_name","N/A"):25s} Score={top.get("match_score",0):3d}/100  Latency={elapsed_ms:.0f}ms')

print(f'\nAvg latency: {np.mean(latencies):.0f}ms | Max: {np.max(latencies):.0f}ms | Min: {np.min(latencies):.0f}ms')
```

## 5. Performance Analysis

### 5.1 Score Distribution Across JDs

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 11))
fig.patch.set_facecolor('#0d1117')
fig.suptitle('RAG Profile Matching - Performance Analysis', color='white', fontsize=15, fontweight='bold')

# Plot 1: Score distributions per JD
ax1 = axes[0, 0]
ax1.set_facecolor('#141929')
for idx, (jd_name, res) in enumerate(all_results.items()):
    scores = [m['match_score'] for m in res['top_matches']]
    ax1.plot(range(1, len(scores)+1), scores, marker='o', label=jd_name.replace('_',' ').title()[:18], color=COLORS[idx % len(COLORS)], linewidth=2, markersize=5)
ax1.set_title('Match Score Curves (Top-K)', color='white', fontweight='bold')
ax1.set_xlabel('Rank', color='#94a3b8')
ax1.set_ylabel('Match Score (0-100)', color='#94a3b8')
ax1.tick_params(colors='#94a3b8')
ax1.legend(fontsize=7, loc='upper right')
for spine in ax1.spines.values(): spine.set_edgecolor('#374151')

# Plot 2: Latency bar chart
ax2 = axes[0, 1]
ax2.set_facecolor('#141929')
jd_labels = [k.replace('_',' ').title()[:15] for k in all_results.keys()]
bars = ax2.bar(jd_labels, latencies, color=COLORS[1], alpha=0.85)
ax2.axhline(np.mean(latencies), color='#f87171', linestyle='--', linewidth=1.5, label=f'Mean: {np.mean(latencies):.0f}ms')
ax2.set_title('Retrieval Latency per JD', color='white', fontweight='bold')
ax2.set_ylabel('Latency (ms)', color='#94a3b8')
ax2.tick_params(colors='#94a3b8', axis='x', rotation=35)
ax2.legend(fontsize=9)
for spine in ax2.spines.values(): spine.set_edgecolor('#374151')

# Plot 3: Top candidate scores per JD
ax3 = axes[1, 0]
ax3.set_facecolor('#141929')
top_scores = [res['top_matches'][0]['match_score'] if res['top_matches'] else 0 for res in all_results.values()]
bar_colors = ['#4ade80' if s >= 70 else '#fbbf24' if s >= 45 else '#f87171' for s in top_scores]
ax3.bar(jd_labels, top_scores, color=bar_colors, alpha=0.9)
ax3.set_title('Top Candidate Score per JD', color='white', fontweight='bold')
ax3.set_ylabel('Match Score (0-100)', color='#94a3b8')
ax3.set_ylim(0, 105)
ax3.tick_params(colors='#94a3b8', axis='x', rotation=35)
for spine in ax3.spines.values(): spine.set_edgecolor('#374151')
green_p = mpatches.Patch(color='#4ade80', label='Strong (>=70)')
yellow_p = mpatches.Patch(color='#fbbf24', label='Moderate (45-69)')
red_p = mpatches.Patch(color='#f87171', label='Weak (<45)')
ax3.legend(handles=[green_p, yellow_p, red_p], fontsize=8)

# Plot 4: Skill coverage scatter
ax4 = axes[1, 1]
ax4.set_facecolor('#141929')
all_scores_flat = [m['match_score'] for res in all_results.values() for m in res['top_matches']]
all_exp_flat = [m['years_experience'] for res in all_results.values() for m in res['top_matches']]
all_skills_cnt = [len(m['matched_skills']) for res in all_results.values() for m in res['top_matches']]
sc = ax4.scatter(all_exp_flat, all_scores_flat, c=all_skills_cnt, cmap='viridis', s=60, alpha=0.7)
plt.colorbar(sc, ax=ax4, label='Matched Skills Count')
ax4.set_title('Score vs Experience (colored by matched skills)', color='white', fontweight='bold')
ax4.set_xlabel('Years of Experience', color='#94a3b8')
ax4.set_ylabel('Match Score', color='#94a3b8')
ax4.tick_params(colors='#94a3b8')
for spine in ax4.spines.values(): spine.set_edgecolor('#374151')

plt.tight_layout()
plt.savefig('performance_analysis.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()
print('Chart saved: performance_analysis.png')
```

### 5.2 Retrieval Accuracy Metrics

```python
# Compute summary metrics table
metrics = []
for jd_name, res in all_results.items():
    tops = res['top_matches']
    scores = [m['match_score'] for m in tops]
    metrics.append({
        'JD': jd_name.replace('_', ' ').title(),
        'Candidates': res['total_candidates_evaluated'],
        'Top Score': scores[0] if scores else 0,
        'Avg Top-5': round(np.mean(scores[:5]), 1) if scores else 0,
        'Latency (ms)': res['retrieval_latency_ms'],
        'JD Skills': len(res.get('jd_skills', [])),
        'Min Exp Req': res.get('min_experience_required'),
    })

metrics_df = pd.DataFrame(metrics)
print('=== Retrieval Performance Metrics ===')
print(metrics_df.to_string(index=False))
print(f'\nOverall avg latency: {metrics_df["Latency (ms)"].mean():.1f}ms')
print(f'Overall avg top score: {metrics_df["Top Score"].mean():.1f}/100')
metrics_df.to_csv('metrics_summary.csv', index=False)
print('Saved: metrics_summary.csv')
```

### 6.1 Deep Dive: ML Engineer JD

```python
print('=== ML Engineer JD - Top 3 Matches ===\n')
for i, m in enumerate(result_ml['top_matches'][:3], 1):
    print(f'Rank {i}: {m["candidate_name"]}')
    print(f'  Score:    {m["match_score"]}/100')
    print(f'  Exp:      {m["years_experience"]} years')
    print(f'  Sections: {m.get("sections_matched", [])}')
    print(f'  Skills:   {m["matched_skills"]}')
    print(f'  Reasoning: {m["reasoning"][:300]}')
    if m.get('relevant_excerpts'):
        print(f'  Excerpt:  {m["relevant_excerpts"][0][:200]}...')
    print()
```

### 6.2 Skills Heatmap: Candidates × JDs

```python
# Build a candidate x JD match score matrix for top candidates
candidate_names = list({m['candidate_name'] for res in all_results.values() for m in res['top_matches'][:5]})
jd_names = list(all_results.keys())

matrix = np.zeros((len(candidate_names), len(jd_names)))
for j, jd_name in enumerate(jd_names):
    score_map = {m['candidate_name']: m['match_score'] for m in all_results[jd_name]['top_matches']}
    for i, cname in enumerate(candidate_names):
        matrix[i, j] = score_map.get(cname, 0)

fig, ax = plt.subplots(figsize=(14, max(6, len(candidate_names) * 0.45)))
fig.patch.set_facecolor('#0d1117')
ax.set_facecolor('#141929')
im = ax.imshow(matrix, cmap='RdYlGn', aspect='auto', vmin=0, vmax=100)
plt.colorbar(im, ax=ax, label='Match Score (0-100)')
ax.set_xticks(range(len(jd_names)))
ax.set_xticklabels([j.replace('_',' ').title()[:15] for j in jd_names], rotation=35, ha='right', color='#94a3b8', fontsize=9)
ax.set_yticks(range(len(candidate_names)))
ax.set_yticklabels(candidate_names, color='#94a3b8', fontsize=9)
ax.set_title('Candidate × JD Match Score Heatmap', color='white', fontweight='bold', fontsize=14)
plt.tight_layout()
plt.savefig('match_heatmap.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()
print('Heatmap saved: match_heatmap.png')
```
